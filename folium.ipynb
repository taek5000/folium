행정동별 색상을 반영한 지도 생성

```python
import pandas as pd
import folium
import random
import seaborn as sns
from matplotlib.colors import to_hex
```

```python
df = pd.DataFrame({'행정동':['역삼동', '삼성동', '논현동', '신림동', '미아동'],
                             '위도':[37.5013, 37.5146, 37.5114, 37.4825, 37.6265],
                             '경도':[127.0396, 127.0561, 127.0216, 126.9290, 127.0250],
                             '색상': ['blue', 'red', 'green', 'orange', 'purple']})
df.head()
```

```python
# 서울 지도 생성
m = folium.Map(location=[37.5665, 126.9780], zoom_start=12, tiles="cartodb positron")
```

```python
# 행정동별 색상으로 원(CircleMarker) 추가
for _, row in df.iterrows():
    folium.CircleMarker(
        location=[row["위도"], row["경도"]],
        radius=10,  # 원 크기
        color=row["색상"],  # 테두리 색
        fill=True,
        fill_color=row["색상"],  # 채우기 색
        fill_opacity=0.6,
        popup=row["행정동"],
    ).add_to(m)
```

```python
# 서울 중심 지도 생성
m = folium.Map(location=[37.5665, 126.9780], zoom_start=12, tiles="cartodb positron")
```

```python
# 행정동별 랜덤 색상 지정 (seaborn 사용하여 부드러운 색상 선택)
color_palette = sns.color_palette("husl", 11)  # 'husl' 컬러맵 사용
hex_colors = [to_hex(color) for color in color_palette]  # 수정된 HEX 변환 함수
random.shuffle(hex_colors)  # 색상 랜덤 배치
```

```python
# FeatureGroup을 사용하여 그룹핑
feature_group = folium.FeatureGroup(name="서울 행정동")
```

```python
# 행정동별 색상 적용 및 표시
for idx, row in df.iterrows():
    folium.CircleMarker(
        location=[row["위도"], row["경도"]],
        radius=10,  # 원 크기 (조절 가능)
        color=hex_colors[idx],  # 테두리 색상
        fill=True,
        fill_color=hex_colors[idx],  # 내부 색상
        fill_opacity=0.7,
        tooltip=f"<b>{row['행정동']}</b>",  # 마우스를 올리면 행정동 이름 표시
    ).add_to(feature_group)
```

```python
# 지도에 FeatureGroup 추가
feature_group.add_to(m)
```

```python
# LayerControl 추가 (여러 개의 Layer를 추가할 경우)
folium.LayerControl().add_to(m)
```

```python
# 원본 데이터 로드
with open("seoul_data.json", "r", encoding="utf-8") as f:
    seoul_data = json.load(f)
```

```python
from pyproj import Transformer

# 좌표 변환기 설정 (EPSG:5179 → EPSG:4326)
transformer = Transformer.from_crs("EPSG:5179", "EPSG:4326", always_xy=True)
```

```python
# 전체 좌표 변환 함수 (X, Y가 반대인 경우 처리)
def transform_coordinates(geometry):
    if geometry["type"] == "Polygon":
        return [[transformer.transform(x, y) for x, y in polygon] for polygon in geometry["coordinates"]]
    elif geometry["type"] == "MultiPolygon":
        return [[[transformer.transform(x, y) for x, y in polygon] for polygon in multi] for multi in geometry["coordinates"]]
    else:
        raise ValueError(f"Unsupported geometry type: {geometry['type']}")
```

```python
# 변환할 GeoJSON 데이터 구조
geojson = {
    "type": "FeatureCollection",
    "features": []
}
```

```python
# 데이터 변환
for feature in seoul_data['features']:
    geojson["features"].append({
        "type": "Feature",
        "geometry": {
            "type": "Polygon",
            "coordinates": transform_coordinates(feature["geometry"])  # 좌표 정보
        },
        "properties": feature["properties"]  # 속성 정보
    })
```

```python
# GeoJSON 저장
with open("boundary.geojson", "w", encoding="utf-8") as f:
    json.dump(geojson, f, ensure_ascii=False, indent=4)

print("GeoJSON 변환 완료: boundary.geojson")
```
